# Network Log Analyzer & Parser

An automated Python script designed to process, parse, and analyze raw network logs, categorize IP address types, and generate automated CSV reports.


## Problem Statement
Dalam administrasi jaringan dan monitoring keamanan, pengelola sistem menangani ribuan data log setiap hari. Memilih IP valid, memisahkan trafik internal (privat) dari koneksi luar (publik), dan menemukan data rusak secara manual membutuhkan waktu lama dan rawan kesalahan.

## Solution & Features
Script Python otomatis ini bertugas untuk:
1. Memvalidasi kelayakan format IPv4 secara otomatis.
2. Memisahkan data IP yang rusak/salah format.
3. Memilih & mengelompokkan IP Privat vs IP Publik sesuai aturan standar jaringan (RFC 1918).
4. Menghitung frekuensi akses tiap IP untuk mendeteksi potensi anomali trafik.
5. Meng-export hasil rekapitulasi data langsung ke file `.csv` (Excel).


## 🛠️ Tech Stack
* **Bahasa:** Python 3.x
* **Library Utama:** `ipaddress` (validasi IP), `csv` (export data)
* **Environment:** Google Colab Mobile

```python
import csv
import ipaddress

# SIMULASI DATA LOG JARINGAN
raw_logs = [
    "192.168.1.50",
    "10.0.0.1",
    "8.8.8.8",
    "256.100.0.1",  # IP Tidak Valid
    "172.16.0.5",
    "1.1.1.1",
    "192.168.1.50",  # Duplikat
    "invalid_ip_address",  # Format Rusak
    "8.8.8.8",
    "10.0.0.1",
]


def analisa_log_jaringan(logs):
    ip_valid_list = []
    ip_invalid_list = []
    ip_privat = []
    ip_publik = []
    frekuensi_ip = {}

    print("=== PROSES ANALISIS LOG JARINGAN ===\n")

    for entry in logs:
        frekuensi_ip[entry] = frekuensi_ip.get(entry, 0) + 1

        try:
            ip_obj = ipaddress.ip_address(entry)
            ip_valid_list.append(entry)

            if ip_obj.is_private:
                ip_privat.append(entry)
            else:
                ip_publik.append(entry)

        except ValueError:
            ip_invalid_list.append(entry)

    print(f"Total Log Diproses : {len(logs)}")
    print(f"IP Valid           : {len(ip_valid_list)}")
    print(f"IP Tidak Valid     : {len(ip_invalid_list)}")
    print(f"IP Privat (Lokal)  : {len(set(ip_privat))}")
    print(f"IP Publik          : {len(set(ip_publik))}\n")

    # EXPORT KE CSV
    nama_file_csv = "rekap_log_jaringan.csv"
    with open(nama_file_csv, mode="w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["IP Address", "Frekuensi Akses", "Status Validasi"])

        for ip, count in frekuensi_ip.items():
            status = "Valid" if ip in ip_valid_list else "Tidak Valid"
            writer.writerow([ip, count, status])

    print(f"✓ Rekap data berhasil di-export ke file: '{nama_file_csv}'")


# Jalankan Program
analisa_log_jaringan(raw_logs)
```
